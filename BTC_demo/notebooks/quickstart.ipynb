# BTC: graph construction, then byster prediction
Run each step separately in the configured BTC environment.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
ROOT = Path.cwd().resolve()
if not (ROOT / "creat_graph.py").is_file():
    ROOT = ROOT.parent
assert (ROOT / "creat_graph.py").is_file()
INPUT = ROOT / "examples/sample_graph_input.h5ad"
GRAPH_DIR = ROOT / "graphs"
OUTPUT_DIR = ROOT / "predictions"
```

## 1. Build graphs
Build graphs in a new or empty GRAPH_DIR.

```python
subprocess.run([sys.executable, str(ROOT / "creat_graph.py"), str(INPUT), "--output-dir", str(GRAPH_DIR), "--target-size", "50"], check=True)
```

## 2. Predict with BTC
Reuse the graphs above. Use a new or empty OUTPUT_DIR.

```python
subprocess.run([sys.executable, str(ROOT / "byster.py"), "--graph-dir", str(GRAPH_DIR), "--output-dir", str(OUTPUT_DIR)], check=True)
```
